```python
# Monthly Budget Planner using Python, Pandas, and NumPy
# Student Name: Aryan Maurya (B.Tech 1st Year)

import pandas as pd
import numpy as np

# 1. Define Monthly Budget Limits (in INR)
budget_data = {
    'Category': ['Food', 'Travel', 'Books & Study', 'Entertainment', 'Personal Care'],
    'Budget_Limit': [2500, 1200, 1500, 800, 500]
}
df_budget = pd.DataFrame(budget_data)

# 2. Record Daily/Monthly Expense Transactions
expense_data = {
    'Date': [
        '2026-10-02', '2026-10-05', '2026-10-07', '2026-10-10',
        '2026-10-12', '2026-10-15', '2026-10-18', '2026-10-20',
        '2026-10-24', '2026-10-28'
    ],
    'Category': [
        'Food', 'Travel', 'Books & Study', 'Entertainment',
        'Food', 'Personal Care', 'Travel', 'Food',
        'Books & Study', 'Entertainment'
    ],
    'Description': [
        'Mess & Canteen', 'Metro Card Recharge', 'Engineering Maths Book', 'Movie Ticket',
        'Snacks & Dinner', 'Toiletries & Haircut', 'Auto & Bus Fare', 'Groceries',
        'Lab Manual & Printouts', 'Online Gaming & Cafe'
    ],
    'Amount': [950, 450, 780, 420, 680, 350, 400, 720, 410, 550]
}
df_expenses = pd.DataFrame(expense_data)

# 3. Statistical Analysis using NumPy
amounts = np.array(df_expenses['Amount'])
total_spent = np.sum(amounts)
average_expense = np.mean(amounts)
highest_expense = np.max(amounts)
lowest_expense = np.min(amounts)
std_deviation = np.std(amounts)

print("==========================================")
print("   MONTHLY BUDGET & EXPENSE SUMMARY       ")
print("==========================================")
print(f"Total Monthly Budget : INR {np.sum(df_budget['Budget_Limit'])}")
print(f"Total Amount Spent   : INR {total_spent}")
print(f"Average Transaction  : INR {average_expense:.2f}")
print(f"Highest Single Spend : INR {highest_expense}")
print(f"Lowest Single Spend  : INR {lowest_expense}")
print(f"Spending Std Dev     : INR {std_deviation:.2f}\n")

# 4. Category-Wise Grouping and Budget Comparison using Pandas
category_summary = df_expenses.groupby('Category')['Amount'].sum().reset_index()
category_summary.rename(columns={'Amount': 'Actual_Spent'}, inplace=True)

# Merge budget limits with actual expenses
report = pd.merge(df_budget, category_summary, on='Category', how='left')
report['Actual_Spent'] = report['Actual_Spent'].fillna(0)
report['Remaining_Balance'] = report['Budget_Limit'] - report['Actual_Spent']

# Calculate spending percentage and status using NumPy
report['Usage_%'] = np.round((report['Actual_Spent'] / report['Budget_Limit']) * 100, 2)
report['Status'] = np.where(report['Remaining_Balance'] < 0, 'Over Budget', 'Within Budget')

print("==========================================")
print("      CATEGORY-WISE BUDGET ANALYSIS       ")
print("==========================================")
print(report.to_string(index=False))

# 5. Export Final Report to CSV File
report.to_csv('Aryan_Monthly_Budget_Report.csv', index=False)
print("\n[Success] Report saved to 'Aryan_Monthly_Budget_Report.csv'")
```
